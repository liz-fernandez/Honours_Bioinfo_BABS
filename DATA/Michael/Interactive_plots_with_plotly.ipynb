### Making graphs in python

There are currently three major tools researchers use in Python to make graphs; Matplotlib, Seaborn and Plotly.

- Matplotlib is the most basic of the tools and the one a lot of people start with
- Seaborn is built on top of Matplotlib and is geared towards creating more visually appealing plots than matplotlib
- Plotly, meanwhile, is a separate tool with a focus on interactive plots

Each of these tools has their own merits and disadvantages, and if you continue to program in Python you'll likely end up using all three. However, today I want to highlight the merits of Plotly in particular.

### Why should I use Plotly over Matplotlib or Seaborn?

As a researcher, most of the visualisations I make are exploratory. I've just received some data and I want to get a rough idea of what it says. Or, alternatively, I want to parse through a lot of data quite quickly. For these purposes, the interactivity plotly offers is essential. I also personally really like how the plots it produces look, in comparison to the other two tools. To demonstrate this, let's examine how a relatively standard volcano plot looks in these different tools.

```python
## Import libraries (i.e. the tools we will be using). Note the "as" allows us to refer to the libraries by a shorthand

# Libraries for working with data
import pandas as pd
import numpy as np

# Libraries for visualising data
import plotly.express as px
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# We'll start by reading in our dataset.
df = pd.read_csv('volcano.csv')

# Let's have a look at what our data contains, using the df.head() command to look at just the first few entries
df.head()
```

### Examining our dataset

The data we are working with in this notebook comes from a 2015 study performed at WEHI in Melbourne. It examined the expression profiles of basal and luminal cells in the mammary gland of virgin, pregnant and lactating mice. The actual findings aren't particularly important for our purposes so don't worry too much if you are unfamiliar with some of the terms used.

Here we will visualize the results of the luminal cells from pregnant vs lactating mice. Our dataset has a number of different columns, but we primarily care about just three of them, namely our genes (SYMBOL), their fold change between conditions (logFC), and the p value significance of these differences (-log10Pvalue). Full dataset details can be found in the appendix.

Volcano plots are scatter plots of p value against fold change. In biology, they are often used to quickly identify biologically significant genes in your data, since outlier data points are both statistically significant and have a large fold change.

Let's take a look at how to make a volcano plot in matplotlib:

```python
# Volcano plots in Matplotlib. We tell it we want to make a scatter plot ("plt.scatter") with the fold change column 
# from our data as our x-axis and the p value as the y axis. If the p value is significant, we'd like it to color the genes.
# The sig color column in our data already contains this info so we pass it to c, matplotlibs color parameter.
# Up regulated genes are red, downregulated blue and not-significant grey
plt.rcParams['figure.figsize'] = [20, 20]
plt.scatter(x=df["logFC"], y=df["-log10Pvalue"],c=df['sig_color'])
```

## Matplotlib: Pros and cons

So in one line we've made our first volcano plot. Hooray! However, it's pretty barebones. There's no axis labels or figure legend. Additionally, we need a separate column in our data ("sig color") to produce this, as unlike seaborn or plotly matplotlib isn't able to convert categorical values to colors by default. It's also quite difficult to separate distinct
points in some of the more dense areas of the plot. Most of these things can still be done in matplotlib, but it will require
writing more complex code. Let's see what the other tools defaults can offer instead.

```python
# Volcano plots in seaborn. You'll notice that the syntax, while a little different, is pretty similar to matplotlib.
# We are need to change plt to sns because we are using a different tool (remember this was the shorthand we used 
# when importing them). scatter is replaced by scatterplot, c by hue, and perhaps most notably by providing the data
# as the first argument we are able to reference the columns by their names alone (i.e. logFC instead of df["logFC"])

sns.scatterplot(data=df, x="logFC", y="-log10Pvalue",hue='sig',palette = sns.color_palette(["blue","red","grey"]))
```

## Seaborn: Pros and cons

With seaborn we can immediately see some improvements. While still only one line of code, we now have axis titles, a figure legend, and distinct data points. Personally, I also think it's significantly more visually appealling. Additionally, some parts such as the palette parameter are completely optional and are just set to keep our color style consistent between comparisons. If you'd like, try deleting the palette part entirely and run the cell again to see what the default colors look like instead. However, this plot can still be improved upon.

```python
# Volcano plots in plotly. Most of the syntax is identical to seaborn, with us providing the dataframe and then referencing
# columns by name. The hover_data is used to specify that we want the gene symbol to be displayed when we hover over a datapoint,
# while the color discrete map functions like palette does in seaborn
fig = px.scatter(df, x="logFC", y="-log10Pvalue",color='sig',hover_data=['SYMBOL'],color_discrete_map={"Up":'red',"Down":'blue',"Not significant": 'grey'})
fig.show("notebook")
```

## Plotly: Pros and cons

Once again, just a single line of code but tons of functionality. As mentioned, the major advantage of plotly is its interactivity. The first way you can see this is by hovering over any dot in the graph. In addition to displaying its exact x and y values, it also tells you the gene it represents, which is very useful when quickly exploring your data. If you try hovering over the blue data point in the top left corner you should see that it is Csn1s2b. This is a calcium-sensitive casein  important in milk production. Since our dataset compares lactating and pregnant mice, it makes sense that this gene is very differentially expressed.

Hover info is not plotly's only perk. If you click on labels in the figure legend you can choose to hide or show them. For instance, you could hide the not significant genes if you just wanted to focus in on your genes that were differentially expressed. You might also notice when your cursor is over the plot a bunch of icons appear in the top right corner. This let's you do a bunch of things, including zooming in and out of the plot and saving the current display as a png. All of this can be quite useful when exploring your data.

Plotly isn't without cons though. Sometimes it can be a little fiddly to get it to display how you want, and it's also less well known than our first two tools so there's significantly less online resources for it.

The last thing we'll do is show how you can reuse these same commands to visualise different datasets, in our case focusing on a subset of around thirty genes that we are interested in

```python
# Let's start by grabbing our list of important genes from a separate csv file
interest = pd.read_csv('volcano_genes.csv')

# We can use this list to create a new dataset containing only genes in our list of interest. We do this using the isin command
df_interest = df[df['SYMBOL'].isin(interest['GeneID'])]

# We can then visualise this new subset just as we did before simply by changing the dataset that we pass in
fig = px.scatter(df_interest, x="logFC", y="-log10Pvalue",color='sig',hover_data=['SYMBOL'],color_discrete_map={"Up":'red',"Down":'blue',"Not significant": 'grey'})
fig.show("notebook")
```

## Appendix

### Dataset details

- ENTREZID: Centralised ID for storing gene info in NCBI
- SYMBOL: Shorthand form of gene name
- GENENAME: Full gene name
- logFC: Logarithmic fold change of gene between pregnant and lactating mice
- AveExpr: Average expression of the gene between the two conditions
- P.Value: Statistical significance of difference in expression between conditions
- adj.P.Val: P.value corrected by number of tests (i.e 15804 genes)
- -log10Pvalue: log 10 of adj.P.Val multiplied by -1 so that the most significant genes have the highest value
- sig and sig_color: -log10Pvalue mapped thresholded for significance. Both directional labels and colors are used due to different requirements of plotting tools
